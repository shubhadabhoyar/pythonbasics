## 2.2 Stacks (LIFO — Last-In, First-Out)
A **Stack** adds and removes items from the same end (the top) in $O(1)$ time. In Python, a standard `list` with `.append()` and `.pop()` works as a fast stack.


```python
def is_valid_parentheses(s: str) -> bool:
    """Classic Stack problem: check whether brackets are properly balanced."""
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    for char in s:
        if char in pairs.values():
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False
    return len(stack) == 0

assert is_valid_parentheses("{[()]}") is True
assert is_valid_parentheses("{[(])}") is False
assert is_valid_parentheses("(((") is False
print("All balanced parentheses tests passed!")
```

## 2.3 Queues & Deques (FIFO — First-In, First-Out)
- **Important:** Never use `list.pop(0)` for a queue—shifting all remaining elements takes $O(n)$ time!
- Always use **`collections.deque`** (doubly-linked block queue), which supports $O(1)$ `.append()` and $O(1)$ `.popleft()`.


```python
from collections import deque

queue = deque(["Task 1", "Task 2", "Task 3"])
queue.append("Task 4")          # Enqueue at rear -> O(1)
next_task = queue.popleft()     # Dequeue from front -> O(1)

print("Processed:", next_task)
print("Remaining queue:", list(queue))
```

## 2.4 Singly Linked Lists
Unlike arrays, a **Linked List** stores elements in separate `Node` objects connected via pointers (`next`), not contiguous memory.
- **Insert / Delete at head:** $O(1)$
- **Search / Access by index:** $O(n)$


```python
class ListNode:
    def __init__(self, val=0, next_node=None):
        self.val = val
        self.next = next_node

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def prepend(self, val) -> None:
        self.head = ListNode(val, self.head)

    def append(self, val) -> None:
        new_node = ListNode(val)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    def reverse(self) -> None:
        """Reverses the linked list in-place in O(n) time and O(1) space."""
        prev = None
        curr = self.head
        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt
        self.head = prev

    def to_list(self) -> list:
        result, curr = [], self.head
        while curr:
            result.append(curr.val)
            curr = curr.next
        return result

ll = SinglyLinkedList()
for x in [1, 2, 3, 4]:
    ll.append(x)
ll.prepend(0)
print("Original:", ll.to_list())
ll.reverse()
print("Reversed:", ll.to_list())
```

## 3.1 Trees & Binary Search Trees (BST)
A **Binary Search Tree (BST)** maintains the invariant that for every node `N`:
- All values in `N.left` are strictly less than `N.val`
- All values in `N.right` are strictly greater than `N.val`

**Operations:** Average $O(\log n)$ on balanced trees; worst-case $O(n)$ if skewed.
- **In-order traversal (`Left -> Root -> Right`)** visits BST nodes in **sorted ascending order**.


```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def insert_bst(root: TreeNode | None, val: int) -> TreeNode:
    if not root:
        return TreeNode(val)
    if val < root.val:
        root.left = insert_bst(root.left, val)
    elif val > root.val:
        root.right = insert_bst(root.right, val)
    return root

def search_bst(root: TreeNode | None, target: int) -> bool:
    if not root:
        return False
    if target == root.val:
        return True
    return search_bst(root.left, target) if target < root.val else search_bst(root.right, target)

def inorder(root: TreeNode | None) -> list[int]:
    return inorder(root.left) + [root.val] + inorder(root.right) if root else []

def preorder(root: TreeNode | None) -> list[int]:
    return [root.val] + preorder(root.left) + preorder(root.right) if root else []

def postorder(root: TreeNode | None) -> list[int]:
    return postorder(root.left) + postorder(root.right) + [root.val] if root else []

# Build BST
bst_root = None
for val in [8, 3, 10, 1, 6, 14, 4, 7]:
    bst_root = insert_bst(bst_root, val)

print("In-order (Sorted): ", inorder(bst_root))
print("Pre-order:         ", preorder(bst_root))
print("Post-order:        ", postorder(bst_root))
print("Search 6:", search_bst(bst_root, 6), "| Search 99:", search_bst(bst_root, 99))
```

## 3.2 Heaps & Priority Queues (`heapq`)
A **Min-Heap** is a complete binary tree where every parent is $\le$ its children, allowing $O(1)$ access to the minimum element and $O(\log n)$ insertion/removal.
- Python's built-in `heapq` operates on standard lists as a **Min-Heap**.
- Tip: To simulate a **Max-Heap**, push negated numbers (`-x`) and negate them back on pop.


```python
import heapq

nums = [15, 3, 9, 1, 22, 7]
heapq.heapify(nums)                 # In-place O(n) heap construction
print("Min element peek:", nums[0])

heapq.heappush(nums, 2)             # O(log n)
smallest = heapq.heappop(nums)      # O(log n) -> returns 1
print("Popped smallest:", smallest)
print("Top 3 largest:", heapq.nlargest(3, nums))
```

## 4.2 Searching: Iterative Binary Search
Binary Search locates a target in a **sorted** list by halving the search interval each step: **$O(\log n)$ time, $O(1)$ space**.


```python
def binary_search(arr: list[int], target: int) -> int:
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

sorted_data = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print("Index of 23:", binary_search(sorted_data, 23))  # 5
print("Index of 99:", binary_search(sorted_data, 99))  # -1
```

## 4.3 Sorting Algorithms
- **Insertion Sort:** $O(n^2)$ worst-case, $O(n)$ on nearly sorted inputs, $O(1)$ auxiliary space.
- **Merge Sort:** Divide-and-conquer algorithm with **guaranteed $O(n \log n)$** runtime and **stability**, using $O(n)$ extra space.
- **Quick Sort:** Divide-and-conquer partitioning around a pivot; $O(n \log n)$ average, $O(n^2)$ worst-case.
- *(Note: Python's built-in `sorted()` uses **Timsort**, a hybrid of Merge Sort and Insertion Sort.)*


```python
def insertion_sort(arr: list[int]) -> list[int]:
    a = arr[:]
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a

def merge_sort(arr: list[int]) -> list[int]:
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return _merge(left, right)

def _merge(left: list[int], right: list[int]) -> list[int]:
    merged = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

def quick_sort(arr: list[int]) -> list[int]:
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

unsorted = [38, 27, 43, 3, 9, 82, 10]
print("Insertion Sort:", insertion_sort(unsorted))
print("Merge Sort:    ", merge_sort(unsorted))
print("Quick Sort:    ", quick_sort(unsorted))
```

## 4.4 Graph Traversal: BFS & DFS
- **Breadth-First Search (BFS):** Uses a **Queue** (`deque`) to explore level-by-level. Finds the **shortest path** (minimum edges) in unweighted graphs in $O(V + E)$ time.
- **Depth-First Search (DFS):** Uses **Recursion** (or an explicit Stack) to explore each branch deeply before backtracking in $O(V + E)$ time.


```python
from collections import deque

def bfs_shortest_path(graph: dict, start: str, goal: str) -> int:
    queue = deque([(start, 0)])
    visited = {start}
    while queue:
        node, dist = queue.popleft()
        if node == goal:
            return dist
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, dist + 1))
    return -1

def dfs_recursive(graph: dict, node: str, visited: set | None = None) -> list[str]:
    if visited is None:
        visited = set()
    visited.add(node)
    order = [node]
    for neighbor in graph[node]:
        if neighbor not in visited:
            order.extend(dfs_recursive(graph, neighbor, visited))
    return order

print("Shortest distance A -> E (BFS):", bfs_shortest_path(graph, "A", "E"))
print("DFS traversal from A:          ", dfs_recursive(graph, "A"))
```

## 4.5 Recursion & Introductory Dynamic Programming (DP)
Dynamic Programming solves problems with **overlapping subproblems** and **optimal substructure** by storing subproblem answers instead of recomputing them exponentially.

**Classic Example — Climbing Stairs:** How many distinct ways can you climb $n$ stairs if you can take either 1 or 2 steps at a time?
- Recurrence: $\text{ways}(n) = \text{ways}(n - 1) + \text{ways}(n - 2)$


```python
from functools import lru_cache

# 1. Top-Down DP (Memoization): O(n) time, O(n) space
@lru_cache(maxsize=None)
def climb_stairs_memo(n: int) -> int:
    if n <= 2:
        return n
    return climb_stairs_memo(n - 1) + climb_stairs_memo(n - 2)

# 2. Bottom-Up DP (Space-Optimized Tabulation): O(n) time, O(1) space
def climb_stairs_dp(n: int) -> int:
    if n <= 2:
        return n
    prev2, prev1 = 1, 2
    for _ in range(3, n + 1):
        prev2, prev1 = prev1, prev1 + prev2
    return prev1

print("Ways to climb 10 stairs (Memo):", climb_stairs_memo(10))
print("Ways to climb 10 stairs (DP):  ", climb_stairs_dp(10))
```

```python
# --- Module 1 Solutions ---
def fizz_buzz_sol(n: int) -> list[str]:
    res = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            res.append("FizzBuzz")
        elif i % 3 == 0:
            res.append("Fizz")
        elif i % 5 == 0:
            res.append("Buzz")
        else:
            res.append(str(i))
    return res

def is_palindrome_sol(s: str) -> bool:
    cleaned = "".join(ch.lower() for ch in s if ch.isalnum())
    return cleaned == cleaned[::-1]

def word_frequencies_sol(sentence: str) -> dict[str, int]:
    counts = {}
    for word in sentence.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts

# --- Module 2 Solutions ---
def two_sum_sol(nums: list[int], target: int) -> list[int]:
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

def has_cycle_sol(head: ListNode) -> bool:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

# --- Module 3 Solutions ---
def max_depth_sol(root: TreeNode | None) -> int:
    if not root:
        return 0
    return 1 + max(max_depth_sol(root.left), max_depth_sol(root.right))

def is_valid_bst_sol(root: TreeNode | None, low=float("-inf"), high=float("inf")) -> bool:
    if not root:
        return True
    if not (low < root.val < high):
        return False
    return is_valid_bst_sol(root.left, low, root.val) and is_valid_bst_sol(root.right, root.val, high)

# --- Module 4 Solutions ---
def first_bad_version_sol(arr: list[bool]) -> int:
    left, right = 0, len(arr) - 1
    ans = -1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid]:
            ans = mid
            right = mid - 1
        else:
            left = mid + 1
    return ans

def coin_change_sol(coins: list[int], amount: int) -> int:
    dp = [float("inf")] * (amount + 1)
    dp[0] = 0
    for a in range(1, amount + 1):
        for c in coins:
            if a - c >= 0:
                dp[a] = min(dp[a], dp[a - c] + 1)
    return dp[amount] if dp[amount] != float("inf") else -1

# Verification Tests
assert fizz_buzz_sol(5) == ["1", "2", "Fizz", "4", "Buzz"]
assert is_palindrome_sol("Race car") is True
assert word_frequencies_sol("to be or not to be") == {"to": 2, "be": 2, "or": 1, "not": 1}
assert two_sum_sol([2, 7, 11, 15], 9) == [0, 1]

n1, n2, n3 = ListNode(1), ListNode(2), ListNode(3)
n1.next, n2.next, n3.next = n2, n3, n2
assert has_cycle_sol(n1) is True

assert max_depth_sol(bst_root) == 4
assert is_valid_bst_sol(bst_root) is True
assert first_bad_version_sol([False, False, True, True, True]) == 2
assert coin_change_sol([1, 2, 5], 11) == 3

print("All Module 1-4 solution assertions passed successfully!")
```
